Reading Subsets of the data

Geometry filter -> the geometry filter only loads data that intersects with the geometry


```python
gdf_mask = gpd.read_file(
    geodatasets.get_path('geoda.nyc')
)
gdf = gpd.read_file(
    geodatasets.get_path('geoda.nyc education'),
    mask = gdf_mask[gdf_mask.name=='Coney Island'],
)
```

Bounding Box filter -> the bounding box filter only loads data that intersexts with the bounding box

```python
bbox = (
    1031051.7879884212, 224272.49231459625, 1047224.3104931959, 244317.30894023244
)
gdf = gpd.read_file(
    geodatasets.get_path('nybb'),
    bbox = bbox
)
```
